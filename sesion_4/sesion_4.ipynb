# Filtrado y ordenamiento de la base de datos

```python
# Cómo traer una columna o varias de nuestro data frame
df_prueba[["id_cliente", "ciudad"]]

# Guardarlo en una nueva base de datos
df_subset = df_prueba[["id_cliente", "ciudad", "edad", "num_pedidos"]].copy()

display(df_subset)
```

```python
# Filtros basados en condiciones:

df_subset_2 = df_prueba[df_prueba["num_pedidos"]>10]

display(df_subset_2)
```

```python
# Filtros basados en condiciones:

df_subset_3 = df_prueba[(df_prueba["num_pedidos"]>10) & (df_prueba["ticket_promedio_miles_cop"] > 100)]

display(df_subset_3)
```

```python
# Filtros basados en condiciones:

df_subset_4 = df_prueba[(df_prueba["ciudad"] == "Barranquilla")]

display(df_subset_4)
```

```python
# | -> or
# & -> and
# ~ -> negación

# Filtros basados en condiciones:

df_subset_5 = df_prueba[(df_prueba["num_pedidos"]>10) | (df_prueba["ticket_promedio_miles_cop"] > 100)]

display(df_subset_5)
```

```python
# ~ -> negación

# Filtros basados en condiciones:

df_subset_6 = df_prueba[ ~ (df_prueba["ciudad"] == "Barranquilla")]

display(df_subset_6)
```

# EDA: Análisis Exploratorio de Datos. (Exploratory Data Analysis)

```python
# Separar los valores numéricos de nuestra base de datos

df_numbers = df_prueba.select_dtypes(include=[np.number])
```

```python
df_numbers
```

```python
stats_num = df_numbers.describe().T
stats_num
```

```python
# Añadir datos faltantes

stats_num['kurtosis'] = df_numbers.kurt()
stats_num['skewness'] = df_numbers.skew()
stats_num['nulos'] = df_numbers.isnull().sum()

display(stats_num)
```

```python
def graficar_variable_escalar(variable_a_analizar):
    # Extraer la variable asegurando que sea numerica y sin valores nulos
    data_var = df_numbers[variable_a_analizar].dropna()

    # Calcular Asimetria y Curtosis
    asim = data_var.skew()
    curt = data_var.kurt()

    # Configurar la grafica
    fig, ax = plt.subplots(figsize=(12, 6))

    # Histograma con KDE (curva de densidad)
    sns.histplot(data_var, kde=True, ax=ax, color='coral', stat='density', bins=30)

    # Añadir titulos y etiquetas
    ax.set_title(f'Distribucion de {variable_a_analizar}\nAsimetria (Skew): {asim:.2f} | Curtosis (Kurt): {curt:.2f}', fontsize=14, pad=15)
    ax.set_xlabel('Valor', fontsize=12)
    ax.set_ylabel('Densidad', fontsize=12)

    # Lineas de referencia para la media y mediana
    ax.axvline(data_var.mean(), color='red', linestyle='dashed', linewidth=2, label=f'Media: {data_var.mean():.2f}')
    ax.axvline(data_var.median(), color='blue', linestyle='solid', linewidth=2, label=f'Mediana: {data_var.median():.2f}')
    ax.legend()

    plt.tight_layout()
    plt.show()
```

```python
graficar_variable_escalar('antiguedad_dias')
```

```python
graficar_variable_escalar('edad')
```

```python

graficar_variable_escalar('gasto_total_miles_cop')
```

```python
graficar_variable_escalar('minutos_sesion_promedio')
```

```python
# Hacer la matriz de correlaciones

matriz_corr = df_numbers.corr()
```

```python
plt.figure(figsize=(10, 8))

sns.heatmap(matriz_corr, 
            annot=True,          # annot=True pone los numeritos dentro de los cuadros
            cmap='coolwarm',     # cmap='coolwarm' usa tonos rojos (positiva) y azules (negativa)
            fmt=".2f",           # Muestra solo 2 decimales
            linewidths=0.5)      # Separa un poco los cuadritos

plt.title('Matriz de Correlación de Variables Numéricas', fontsize=16)
plt.tight_layout()
plt.show()
```
