```python
download_path = kagglehub.competition_download('gsis-ml-society-section-b')
```

```python
train_base = pd.read_csv(os.path.join(download_path, 'train.csv'))
test_base = pd.read_csv(os.path.join(download_path, 'test.csv'))
```

```python
train_base.head()
```

```python
train_base.drop(['id'], axis=1, inplace=True)
test_base.drop(['id'], axis=1, inplace=True)
```

```python
train_base['Sex'] = train_base['Sex'].astype('category').cat.codes
test_base['Sex'] = test_base['Sex'].astype('category').cat.codes
```

```python
X = train_base.drop(['Age'], axis=1)
y = train_base['Age']

display(X.head())
display(y.head())
```

```python
rf = RandomForestRegressor(n_estimators=500, max_features=0.5, min_samples_leaf=2, n_jobs=-1)
```

```python
rf.fit(X, y)
```

```python
y_test = rf.predict(test_base)
```

```python
sub = pd.DataFrame({'id': test_base['id'], 'Age': y_test})
```

```python
sub.to_csv('submission.csv', index=False)
```
