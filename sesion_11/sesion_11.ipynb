```python
from typing import Literal
```

```python
documents = SimpleDirectoryReader(input_files = ['El-jardin-de-senderos-que-se-bifurcan-JL-Borges.pdf']).load_data()

print(f"Se cargaron:  {len(documents)} páginas")
```

```python
nodes = splitter.get_nodes_from_documents(documents)

print(f"Total de nodos: {len(nodes)}")
```

Crear la ontología

```python
TIPOS_ENTIDAD = ["PERSONA", "LUGAR", "ORGANIZACION", "OBRA", "OBJETO", "EVENTO", "CONCEPTO"]

REGLAS = {
    "ASESINA_A":       (["PERSONA"], ["PERSONA"]),
    "PERSIGUE_A":      (["PERSONA"], ["PERSONA"]),
    "DESCIENDE_DE":    (["PERSONA"], ["PERSONA"]),
    "ENEMIGO_DE":      (["PERSONA", "ORGANIZACION"], ["PERSONA", "ORGANIZACION"]),
    "TRABAJA_PARA":    (["PERSONA"], ["PERSONA", "ORGANIZACION"]),
    "VISITA_A":        (["PERSONA"], ["PERSONA", "LUGAR"]),
    "VIAJA_A":         (["PERSONA"], ["LUGAR"]),
    "VIVE_EN":         (["PERSONA"], ["LUGAR"]),
    "UBICADO_EN":      (["LUGAR", "OBJETO", "ORGANIZACION", "EVENTO"], ["LUGAR"]),
    "AUTOR_DE":        (["PERSONA"], ["OBRA"]),
    "ESTUDIA":         (["PERSONA"], ["OBRA", "CONCEPTO", "PERSONA"]),
    "POSEE":           (["PERSONA", "ORGANIZACION"], ["OBJETO", "OBRA"]),
    "PARTICIPA_EN":    (["PERSONA", "ORGANIZACION"], ["EVENTO"]),
    "REVELA":          (["PERSONA", "OBRA", "OBJETO"], TIPOS_ENTIDAD),
    "SIMBOLIZA":       (["OBRA", "OBJETO", "LUGAR", "EVENTO"], ["CONCEPTO"]),
    "RELACIONADO_CON": (TIPOS_ENTIDAD, TIPOS_ENTIDAD),
}

Entidades = Literal[tuple(TIPOS_ENTIDAD)]
Relaciones = Literal[tuple(REGLAS)]
ESQUEMA_VALIDACION = [
    (sujeto, relacion, objeto)
    for relacion, (sujetos, objetos) in REGLAS.items()
    for sujeto in sujetos
    for objeto in objetos
]

print(f"{len(TIPOS_ENTIDAD)} tipos de entidad, {len(REGLAS)} relaciones, "
      f"{len(ESQUEMA_VALIDACION)} combinaciones válidas")
```

Extraer la información con IA

```python
from llama_index.core.indices.property_graph import SchemaLLMPathExtractor
from llama_index.core.types import PydanticProgramMode

llm_extraction = OpenAI(
    model = "gpt-4o-mini",
    temperature = 0,
    pydantic_program_mode = PydanticProgramMode.FUNCTION
)
```

```python
CONTEXTO_OBRA = (
    "Cuento 'El Jardin de los senderos que se bifurcan' de Jorge Luis Borges."
    "Es la declaración de Yu Tsun, cuando el narrador dice 'yo', se refiere a Yu Tsun."
)

PROMPT_EXTRACCION = (
    "Contexto: " + CONTEXTO_OBRA + "\n\n"
    "Extrae un grafo de conocimiento del siguiente fragmento según el esquema dado, "
    "con un máximo de {max_triplets_per_chunk} tripletas.\n"
    "Reglas:\n"
    "- Escribe los nombres en español, en su forma canónica y completa (por ejemplo "
    "'Richard Madden' en lugar de 'Madden' o 'el capitán'), y usa siempre el mismo "
    "nombre para la misma entidad.\n"
    "- Si un mismo nombre designa entidades distintas (por ejemplo, una persona y un "
    "lugar), distínguelas añadiendo el tipo entre paréntesis.\n"
    "- Rellena la propiedad 'descripcion' de cada entidad y de cada relación con una "
    "frase breve basada en el texto.\n"
    "- Extrae solo hechos que aparezcan en el fragmento.\n"
    "-------\n"
    "{text}\n"
    "-------\n"

)


extractor = SchemaLLMPathExtractor(
    llm = llm_extraction,
    extract_prompt = PROMPT_EXTRACCION,
    possible_entities=Entidades,
    possible_relations=Relaciones,
    kg_validation_schema=ESQUEMA_VALIDACION,
    possible_entity_props=[("descripcion", "quién o qué es la entidad, en una frase")],
    possible_relation_props=[("descripcion", "qué ocurre entre sujeto y objeto, en una frase")],
    strict = True,
    max_triplets_per_chunk = 15,
    num_workers=4
)
```

```python
from llama_index.core import PropertyGraphIndex, StorageContext, load_index_from_storage
from llama_index.core.schema import TextNode

PERSIST_DIR="./grafo_cuento"
RECONSTRUIR_GRAFO = False

if os.path.exists(PERSIST_DIR) and not RECONSTRUIR_GRAFO:
    pg_index = load_index_from_storage(StorageContext.from_defaults(persist_dir=PERSIST_DIR))
    print(f"Grafo cargado desde {PERSIST_DIR}")
else:
    chunks_grafo = [
        TextNode(text=n.text, metadata={"pagina": n.metadata.get("page_label")})
        for n in nodes
    ]
    pg_index = PropertyGraphIndex(
        nodes=chunks_grafo,
        kg_extractors=[extractor],
        embed_model=embed_model,  # embeddings de las entidades -> búsqueda semántica en el grafo
        show_progress=True,
    )
    pg_index.storage_context.persist(persist_dir=PERSIST_DIR)
    print(f"Grafo construido y guardado en {PERSIST_DIR}")
```

```python
import networkx as nx
import pandas as pd
from llama_index.core.graph_stores.types import EntityNode

tripletas = [
    {
        "sujeto": s.name,
        "tipo_sujeto": s.label,
        "relacion": r.label,
        "objeto": o.name,
        "tipo_objeto": o.label,
        "descripcion": r.properties.get("descripcion", ""),
    }
    for s, r, o in pg_index.property_graph_store.graph.get_triplets()
    if isinstance(s, EntityNode) and isinstance(o, EntityNode)
]
assert tripletas, "El grafo quedó vacío: revisa la API key, el esquema o el prompt de extracción."

descripciones = {
    nodo.name: nodo.properties.get("descripcion", "")
    for nodo in pg_index.property_graph_store.graph.nodes.values()
    if isinstance(nodo, EntityNode)
}

G = nx.DiGraph()
for t in tripletas:
    for nombre, tipo in [(t["sujeto"], t["tipo_sujeto"]), (t["objeto"], t["tipo_objeto"])]:
        G.add_node(nombre, tipo=tipo, descripcion=descripciones.get(nombre, ""))
    if G.has_edge(t["sujeto"], t["objeto"]):  # varias relaciones entre el mismo par
        G[t["sujeto"]][t["objeto"]]["relacion"] += " / " + t["relacion"]
        G[t["sujeto"]][t["objeto"]]["descripcion"] += " / " + t["descripcion"]
    else:
        G.add_edge(t["sujeto"], t["objeto"], relacion=t["relacion"], descripcion=t["descripcion"])

df_tripletas = pd.DataFrame(tripletas).sort_values(["sujeto", "relacion"], ignore_index=True)
print(f"Entidades: {G.number_of_nodes()} | Relaciones: {len(tripletas)}")
df_tripletas.head(30)
```

```python
comunidades = sorted(
    nx.community.louvain_communities(G.to_undirected(), seed=42), key=len, reverse=True
)
for i, miembros in enumerate(comunidades):
    for n in miembros:
        G.nodes[n]["comunidad"] = i
    print(f"Comunidad {i} ({len(miembros)} entidades): {', '.join(sorted(miembros))}")
```

```python
from pyvis.network import Network

net = Network(height="800px", width="100%", directed=True, cdn_resources="remote")
for n, datos in G.nodes(data=True):
    net.add_node(
        n, label=n, group=datos["comunidad"], size=10 + 3 * G.degree(n),
        title=f"{datos['tipo']} · comunidad {datos['comunidad']}\n{datos['descripcion']}",
    )
for s, o, datos in G.edges(data=True):
    net.add_edge(s, o, label=datos["relacion"], title=datos["descripcion"])

net.write_html("grafo_cuento.html")
print("Grafo interactivo guardado en grafo_cuento.html")
```
