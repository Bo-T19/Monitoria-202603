#### Aplicación estadística

```python
lista_edades = [22, 25, 24, 25, 25, 26, 30, 32, 40, 75, 25, 24, 19]
```

```python
# Funciones aplicables

#Suma de los elementos de la lista
sum(lista_edades)
print(f"La suma de los elementos de la lista es {sum(lista_edades)}")

#Mayor valor de la lista
max(lista_edades)
print(f"El mayor valor de la lista es {max(lista_edades)}")

#Menor valor de la lista
min(lista_edades)
print(f"El menor valor de la lista es {min(lista_edades)}")

# Longitud de la lista
len(lista_edades)
print(f"La longitud de la lista es {len(lista_edades)}")
```

```python
mediana = statistics.median(lista_edades)
print(f"La mediana es {mediana}")

media = sum(lista_edades)/len(lista_edades)
print(f"La media es {media}")

desviscion_estandar = statistics.stdev(lista_edades)
print(f"La desviación estándar es {desviscion_estandar}")

asimetria = skew(lista_edades)
print(f"La asimetría es {asimetria}")

curtosis = kurtosis(lista_edades)
print(f"La curtosis es {curtosis}")
```

## Bucles

### FOR
Usamos el bucle for para recorrer sobre una lista o una coleccion que conocemos a priori. Es para repetir un procedimiento por un número conocido de iteraciones.

```python
# Para repetir un proceso determinado número de veces

for i in range(10):
    print("Hola")
```

```python
lista_imprimir = [1, 3, 5, 9]

for i in lista_imprimir:
    print(i**2)
```

### WHILE
A diferencia del for, el while se ejecuta "mientras" una condición sea verdadera, no sabemos a priori cuántas veces se va a iterar.

```python
# Simulación financiera
capital = 1000
meta = 2000
tasa_crecimiento_mes = 0.05
meses_transcurridos = 0

while capital < meta:
  #1. Calcular el nuevo valor
  ganancia = capital * tasa_crecimiento_mes
  capital = capital + ganancia

  #2. Actualizar el contador de meses
  meses_transcurridos = meses_transcurridos + 1

  #3. Mostrar el resultado:
  print (f"En el mes {meses_transcurridos} tienes $ {capital:.2f} dólares")

print ("Felicitaciones, has llegado al objetivo")
```

```python
# Cómo sería con un for
capital = 1000
tasa_crecimiento_mes = 0.05
meses_transcurridos = 5

for i in range(meses_transcurridos):
    ganancia = capital * tasa_crecimiento_mes
    capital = capital + ganancia
    print (f"En el mes {i+1} tienes $ {capital:.2f} dólares")

```

## Funciones: Type Hints y Defaults

1. Type hints: Le indico al usuario qué tipo de datos esperamos que la función reciba, y le indicamos también qué tipo de dato va a devolver la función.

2. Valores por defecto: Podemos asignar un valor a un parámetro en la creación de la función. Si el usuario no especifca el valor al llamarla, Python usará ese valor por defecto.

```python
def calcular_promedio_estudiante(nombre_estudiante: str, notas: list, bono: float = 0.2) -> float:
  """
  Calcula el promedio de un estudiante, incluyendo un bono adicional. Imprime el valor del promedio.

  Parámetros:
  - nombre_estudiante (str): El nombre del estudiante.
  - notas (list): Una lista de números que representan las notas del estudiante.
  - bono (float): El monto del bono que se le aplica al estudiante. Por defecto es 0.2

  Devuelve:
  - promedio_con_bono (float): El promedio del estudiante, incluyendo el bono.
  """

  promedio = sum(notas)/len(notas)
  promedio_con_bono = promedio + bono

  print(f"El promedio del estudiante {nombre_estudiante} es {promedio_con_bono:.2f}")

  return promedio_con_bono
```

```python
calcular_promedio_estudiante("Juan", [5, 4, 4, 5, 3, 4, 5])
```

## Ejemplo:



```python
def resumen_estadistico(datos: list) -> dict:
  """
  Calcular la estadística descriptiva de un conjunto de datos.

  Parámetros:
  - datos (list): Una lista de números que representan los datos.

  Retorna:
  - dict: un diccionario con el conteo, media, mediana, desviación estándar, asimetría, curtosis, mínimo, máximo
  """

  #1. Verificar que el usuario pase datos correctos
  if len(datos) == 0:
    print ("Por favor envía una lista válida")
    return

  #2. Armar el diccionario con los resultados
  resultados = {
  "Conteo (N)": len(datos),
  "Media" : round(statistics.mean(datos),2),
  "Mediana" : statistics.median(datos),
  "Desviación estándar" : round(statistics.stdev(datos),2),
  "Asimetría" : round(skew (datos),2),
  "Curtosis" : round(kurtosis(datos),2),
  "Mínimo" : min(datos),
  "Máximo" : max(datos)
  }

  return resultados
```

```python
#Probar la función

lista_prueba = [170, 180, 190, 160, 190, 160, 190, 160, 190, 150, 198, 166, 158, 157, 159, 160, 170, 181 ]
```

```python
resumen_estadistico(lista_prueba)
```

```python
resumen_estadistico([])
```
