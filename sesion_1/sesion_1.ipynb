# Operaciones aritméticas

```python
# Operaciones
a = 10
b = 3

# Suma
suma = a + b
print("Suma:", suma)
# Resta
resta = a - b
print ("Resta:", resta)

# Multiplicación
multiplicacion = a * b
print("Multiplicación:", multiplicacion)

# División
division = a / b
print("División:", division)
```

```python
# División entera
division_entera = a // b
print("División entera:", division_entera)
```

```python
# Módulo 
modulo = a % b
print("Módulo:", modulo)
```

```python
# Potencia
potencia = a ** b
print("Potencia:", potencia)
```

```python
# Operación matemática más compleja con scipy

from scipy import integrate
x2 = lambda x: x**2
integrate.quad(x2, 0, 4)
```

## Booleans

Comparar dos enunciados entre sí y decidir su veracidad

Comparadores: >, <, <=, >=, ==, !=

```python
nota_1 = 4.5
nota_2 = 3.7
nota_3 = 2.0

perdio_la_materia_1 = nota_1 <= 3.0
perdio_la_materia_2 = nota_2 <= 3.0
perdio_la_materia_3 = nota_3 <= 3.0

print("Perdió la materia 1:", perdio_la_materia_1)
print("Perdió la materia 2:", perdio_la_materia_2)
print("Perdió la materia 3:", perdio_la_materia_3)
```

# Funciones y condicionales

```python
# Función para calcular el promedio de notas
def promedio(a,b,c):
    return (a + b + c) / 3
```

```python
promedio_estudiante_1=promedio(4, 0, 1)
promedio_estudiante_2=promedio(4, 4, 5)
```

```python
if promedio_estudiante_1 >= 3.0:
    print("El estudiante 1 aprobó con un promedio de:", promedio_estudiante_1)
else:
    print("El estudiante 1 no aprobó con un promedio de:", promedio_estudiante_1)
```

```python
resultado_partido = "1-2"
```

```python
# Resumen de lo visto con un ejemplo: 

goles_local = resultado_partido[0]
goles_visitante = resultado_partido[2]

if int(goles_local) > int(goles_visitante):
    print("El equipo local ganó con un marcador de:", resultado_partido)
elif int(goles_local) < int(goles_visitante):
    print("El equipo visitante ganó con un marcador de:", resultado_partido)
else:
    print("El partido terminó en empate con un marcador de:", resultado_partido)
```
